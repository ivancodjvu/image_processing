# Image Analysis

Starter notebook for microscopy image processing using Python and JupyterLite.

This notebook:
- loads a TIFF or other image,
- displays the image,
- reports basic image information,
- provides a starting point for further image-processing steps.

## 1. Install required packages

In JupyterLite, Python packages can be installed directly in the browser.

```python
import piplite
await piplite.install(['numpy', 'matplotlib', 'scikit-image', 'tifffile'])
```

## 2. Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import tifffile
from skimage import exposure, filters
```

## 3. Load an image

Place your image in the same JupyterLite folder as this notebook and change the filename below.

```python
filename = 'example_image.tif'

image = tifffile.imread(filename)

print('Shape:', image.shape)
print('Data type:', image.dtype)
print('Minimum intensity:', np.min(image))
print('Maximum intensity:', np.max(image))
print('Mean intensity:', np.mean(image))
```

## 4. Display the image

```python
plt.figure(figsize=(7, 7))
plt.imshow(image, cmap='gray')
plt.axis('off')
plt.title('Original image')
plt.show()
```

## 5. Normalize the image

```python
image_normalized = exposure.rescale_intensity(
    image.astype(np.float32),
    out_range=(0, 1)
)

plt.figure(figsize=(7, 7))
plt.imshow(image_normalized, cmap='gray')
plt.axis('off')
plt.title('Normalized image')
plt.show()
```

## 6. Example: Gaussian filtering

```python
filtered = filters.gaussian(image_normalized, sigma=1)

plt.figure(figsize=(7, 7))
plt.imshow(filtered, cmap='gray')
plt.axis('off')
plt.title('Gaussian-filtered image')
plt.show()
```
